Section - A Concept Application

Q1. Data Science Lifecycle

-Problem Definition: The goal is to predict which restaurant zones will have peak demand between 7–9 PM. -Data Collection: Collect two years of order data, including order time, restaurant zone, and order volume. -Data Preparation: Clean missing values, duplicates, incorrect entries, and inconsistent data formats. -Exploration: Analyse demand patterns by zone, day, and time using charts and summary statistics. -Modelling: Train a suitable machine learning model using historical order data to predict peak demand. -Evaluation: Compare predictions with actual demand and measure how accurately the model performs. -Generative AI can speed up Data Preparation by suggesting cleaning code, but incorrect AI-generated code could modify useful data. -It can also help with Exploration by suggesting charts and patterns, but its interpretations must be checked by a human.

Q2. Git Branching Strategy

-A Git branching strategy allows each developer to work separately without directly changing the working main branch. -First, each developer gets the latest code using git pull origin main. -They create their own branch using git checkout -b feature/data-cleaning. -After completing their work, they use git add . and git commit to save their changes. -They push the branch using git push -u origin feature/data-cleaning. -The developer then creates a Pull Request from the feature branch to the main branch for review. -A merge conflict can occur if two developers edit the same lines of the same file. -The conflict is resolved manually, the code is tested, and the corrected changes are committed before completing the merge.
